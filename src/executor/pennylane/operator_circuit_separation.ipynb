```python
import pennylane as qml
from pennylane import numpy as pnp

# Define a device with 1 qubit
dev = qml.device("default.qubit", wires=1)


# Define the parameterized quantum circuit
def circuit(params):
    qml.RX(params[0], wires=0)
    qml.RY(params[1], wires=0)


# Define the parameterized observable
def observable(obs_params):
    return qml.expval(qml.Hamiltonian(obs_params, [qml.PauliZ(0), qml.PauliX(0)]))


# Create a QNode that returns the expectation value of the parameterized observable
@qml.qnode(dev, interface="autograd")
def qnode(*args):
    circuit(args[0])
    return observable(args[1])


# Example parameter values
params = pnp.array([0.5, 0.7], requires_grad=True)
obs_params = pnp.array([1.0, 0.5], requires_grad=True)


print(qnode(params, obs_params))

# Compute gradients
grad_fn = qml.jacobian(qnode, argnum=[0, 1])
grad_params, grad_obs_params = grad_fn(params, obs_params)

print("Gradient w.r.t. circuit parameters:", grad_params)
print("Gradient w.r.t. observable parameters:", grad_obs_params)
```
